# Set up our working environment

```python
# Import required libraries
import pandas as pd
import os
```

```python
# Create directory folders to store our data
dirname = os.getcwd()

raw_data = f"{dirname}/data/raw/"
transformed_data = f"{dirname}/data/transformed/"
refined_data = f"{dirname}/data/refined/"

paths = [raw_data, transformed_data, refined_data]

for path in paths:
    if not os.path.exists(path):
        os.makedirs(path)
```


# Import and store raw data
## SIPRI dataset
Let us first import the dataset.

```python
# Define SIPRI URL for military expenditure data
# This URL points to the SIPRI website where the data is hosted
sipri_url = "https://www.sipri.org/sites/default/files/SIPRI-Milex-data-1949-2024_2.xlsx"
```

Because there are many worksheets inside our excel fie, we will need to:
- Select the individual worksheets we need
- Iterate through each worksheet.
  - Store it in a DataFrame
  - Export it as individual `.csv` files

```python
# Read get a list of individual worksheets
sipri_worksheets = pd.ExcelFile(sipri_url)
sipri_worksheets.sheet_names
```

Ok! We will use:
- `Constant (2022) US$` - table 4
- `Current US$` - table 5
- `Share of GDP` - table 6
- `Share of Govt spending` - table 8

```python
# Create list of table numbers
table_numbers = [4, 5, 6, 8]

for table in table_numbers:
    sipri_raw = pd.read_excel(sipri_url, sheet_name=table)     # Read the data
    sipri_raw.to_csv(f"{raw_data}sipri_data_raw_tb_{table}.csv", index=False)
```
